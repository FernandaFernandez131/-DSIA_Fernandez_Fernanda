Importación de librerías

```
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

datos = np.array([10, 20, 30, 40, 50])

print(datos)
```

Dataset simple con Pandas

```
datos = {
    "estudiante": ["Ana", "Juan", "Carla", "Pedro", "Lucia"],
    "nota": [8, 6, 9, 5, 7]
}

df = pd.DataFrame(datos)


print(df)


print("Promedio:", df["nota"].mean())


print("Nota máxima:", df["nota"].max())


print("Nota mínima:", df["nota"].min())
```

Criterio de selección de herramientas

Pandas → porque permite leer archivos CSV y trabajar con datos organizados en tablas\.
Seaborn → porque permite crear gráficos estadísticos, como gráficos de distribución\.
Scikit\-learn → porque ofrece herramientas para crear y entrenar modelos de Machine Learning\.
NumPy → porque permite realizar operaciones matemáticas y trabajar eficientemente con matrices y arrays\.
